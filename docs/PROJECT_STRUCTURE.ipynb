# Структура проекта

```text
notebooks/
    00_Data.ipynb
    01_EDA.ipynb
    02_EdgeMTSC_pairwise.ipynb
    03_ChannelTokenFormer.ipynb
data/        -- архив UCI, подготовленные данные, маски, metadata
results/     -- CSV, конфигурации, checkpoint, train-статистики
figures/     -- графики EDA, обучения, сравнения
docs/        -- структура и протокол
REPORT.md    -- методология и автоматически сформированные результаты
```

Весь код экспериментов находится в ноутбуках. Дополнительных локальных модулей нет.
Таблицы и отчет создаются после запуска, входной архив скачивается автоматически.
