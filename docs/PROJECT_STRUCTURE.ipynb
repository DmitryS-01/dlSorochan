# Структура проекта

```text
.
├── notebooks/
│   ├── 00_EDA.ipynb
│   ├── 01_EdgeMTSC.ipynb
│   ├── 02_EdgeMTSC_pairwise.ipynb
│   └── 03_Transformer.ipynb
├── docs/
│   ├── PROJECT_STRUCTURE.ipynb
│   ├── EXPERIMENT_SETUP.ipynb
│   └── assets/
│       └── edgemtsc_architecture.png
├── data/                  # появляется после 00_EDA, в git не идет
├── figures/               # EDA, learning curves, confusion matrices, сравнения
├── results/               # json/csv с численными результатами
├── REPORT.md              # 03_Transformer обновляет его реальными цифрами
├── requirements.txt
├── .gitignore
├── LICENSE
└── README.md
```

Никакого `src/` и `run_all.py` нет -- весь код конкретного эксперимента лежит прямо в соответствующем ноутбуке.

## Что создает каждый ноутбук

`00_EDA.ipynb`:
- скачивает raw UCI HAR с Hugging Face;
- делает EDA;
- сохраняет `data/uci_har_raw.npz`;
- сохраняет EDA-графики.

`01_EdgeMTSC.ipynb`:
- обучает standard EdgeMTSC;
- делает 3-fold subject-wise CV;
- запускает маленькую IMP / multiscale ablation;
- считает test metrics, params, train time, latency;
- сохраняет графики и `results/01_*`.

`02_EdgeMTSC_pairwise.ipynb`:
- повторяет тот же протокол с pairwise temporal matrices;
- визуализирует силу связей и самую сильную temporal matrix;
- сохраняет `results/02_*`.

`03_Transformer.ipynb`:
- обучает parameter-matched Transformer;
- строит attention sanity check;
- собирает итоговое сравнение всех моделей;
- обновляет `REPORT.md` реальными результатами.
